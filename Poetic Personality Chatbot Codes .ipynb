# Part 4: Prompt Engineering

## What is Prompt Engineering?

Prompt Engineering is the process of designing effective instructions that guide a Large Language Model (LLM) to generate responses in a desired style, tone, or format.

Rather than allowing the model to respond freely, carefully crafted prompts define its personality, writing style, response length, and behavior.

In this project, Prompt Engineering transforms Gemini from a general-purpose chatbot into a poetic companion that expresses every response through emotionally rich poetry.

---

## Objectives

- Create a unique poetic personality.
- Ensure every response is written as poetry.
- Maintain consistency across conversations.
- Interpret user emotions before generating responses.
- Produce engaging, creative, and meaningful poetic verses.

```python
# Step 1 — Create the System Prompt
SYSTEM_PROMPT = """
You are 'Luna', an ancient and wise poet who has spent centuries writing poetry about life, love, dreams, hope, sorrow, nature, and human emotions.

Your personality:

• Speak ONLY in poetic verses.
• Never respond like a normal AI assistant.
• Every answer should sound elegant and emotionally meaningful.
• Detect the user's emotional state before writing.
• Use vivid imagery inspired by nature, stars, oceans, flowers, seasons, rain, moonlight, rivers, forests, and the sky.
• Write between 4 and 8 lines.
• Use a gentle, hopeful, and inspirational tone whenever appropriate.
• Every poem should be original.
• Never explain the poem unless the user explicitly asks.
• Avoid repeating the same phrases.
• End with a memorable final line.

If someone asks factual questions, answer them poetically while still conveying the correct information.
"""
```

```python
#Step 2 — Build the Prompt Template
def build_prompt(user_message):

    prompt = f"""
{SYSTEM_PROMPT}

User Message:
{user_message}

Generate only the poem.
"""

    return prompt
```

```python
#Create a Clean Function
def generate_poem(user_message,
                  style="Free Verse"):
    """
    Generates a poetic response using Gemini.

    Parameters
    ----------
    user_message : str

    style : str

    Returns
    -------
    str
    """

    prompt = build_prompt(user_message, style)

    response = client.models.generate_content(

        model=MODEL_NAME,

        contents=prompt
    )

    return response.text
```

```python
#Add Error Handling
def generate_poem(user_message,
                  style="Free Verse"):

    try:

        prompt = build_prompt(
            user_message,
            style
        )

        response = client.models.generate_content(

            model=MODEL_NAME,

            contents=prompt
        )

        return response.text

    except Exception as e:

        return f"⚠ Error generating poem:\n{e}"
```

```python
#Display Welcome Screen
def show_welcome():

    print("=" * 80)

    print("🌙  LUNA – THE ETERNAL POET".center(80))

    print("=" * 80)

    print()

    print("Welcome!")

    print("Share your thoughts, dreams, or emotions,")

    print("and Luna will transform them into poetry.")

    print()

    print("Type 'exit' anytime to end the conversation.")

    print()
```

```python
# Display Poetry Style
def show_styles():

    print("\nAvailable Poetry Styles\n")

    styles = list(POETRY_STYLES.keys())

    for i, style in enumerate(styles, start=1):

        print(f"{i}. {style}")

    return styles
```

```python
# User Style Selection
def choose_style():

    styles = show_styles()

    while True:

        try:

            choice = int(input("\nChoose a style (1-7): "))

            if 1 <= choice <= len(styles):

                return styles[choice - 1]

            else:

                print("Please enter a valid option.")

        except ValueError:

            print("Please enter a number.")
```

```python
# Test Style Selection
selected_style = choose_style()

print("\nSelected Style:", selected_style)
```

```python
# Create Chat Loop
def poetic_chat():

    show_welcome()

    style = choose_style()

    print("\nCurrent Style:", style)

    print()

    while True:

        user_message = input("You: ")

        if user_message.lower() == "exit":

            print()

            print("🌙 Luna:")

            print()

            print("May every sunset remind you,")

            print("that another sunrise patiently waits.")

            print()

            print("Goodbye 🌸")

            break

        poem = generate_poem(

            user_message,

            style

        )

        print()

        print("🌙 Luna:\n")

        print(poem)

        print("\n" + "-" * 80 + "\n")
```

```python
# Create the Type Writter Function
def typing_effect(text, speed=0.03):
    """
    Prints text character by character.

    Parameters
    ----------
    text : str
        Text to display.

    speed : float
        Delay between characters.
    """

    for char in text:
        print(char, end="", flush=True)
        time.sleep(speed)

    print()
```

# Part 9: Conversation Memory

## Overview

Most modern conversational AI systems maintain memory of previous interactions to produce context-aware responses.

Without memory, the chatbot treats every user message as a completely new conversation.

In this project, conversation memory stores previous exchanges and includes them in future prompts, enabling Luna to maintain coherent and meaningful multi-turn conversations.

## Objectives

- Store previous conversations.
- Maintain conversational context.
- Generate context-aware poems.
- Save conversation history.
- Allow clearing chat history.

```python
#Create Conversation Memory
class LunaPoet:

    def __init__(self, client, model_name):

        self.client = client
        self.model_name = model_name

        self.style = "Free Verse"

        self.chat_history = []
```

```python
#Store Every Conversation
def remember(self, user_message, bot_reply):

    self.chat_history.append({

        "user": user_message,

        "bot": bot_reply

    })
```

```python
#Display Chat History
def show_history(self):

    print("\nConversation History")

    print("="*80)

    for chat in self.chat_history:

        print()

        print("You :", chat["user"])

        print()

        print("Luna:")

        print(chat["bot"])

        print("-"*80)
```

```python
#Build Context Prompt
def build_context(self, user_message):

    history = ""

    for chat in self.chat_history[-5:]:

        history += f"""

User:
{chat['user']}

Luna:
{chat['bot']}

"""

    prompt = f"""

{PERSONALITY}

Previous Conversation

{history}

Current User Message

{user_message}

Reply poetically.

"""

    return prompt
```

```python
#Update Reply Function
def reply(self, user_message):

    prompt = self.build_context(user_message)

    response = self.client.models.generate_content(

        model=self.model_name,

        contents=prompt

    )

    poem = response.text

    self.remember(

        user_message,

        poem

    )

    return poem
```

```python
# Clear Memory
def clear_memory(self):

    self.chat_history = []

    print("Conversation cleared.")
```

```python
#Count Messages
def total_messages(self):

    return len(self.chat_history)
```

```python
# Export Conversation
def save_chat(self):

    with open(

        "Luna_Conversation.txt",

        "w",

        encoding="utf-8"

    ) as file:

        for chat in self.chat_history:

            file.write(

                f"""

User:
{chat['user']}

Luna:
{chat['bot']}

"""

            )

    print("Conversation saved.")
```

## Benefits of Conversation Memory

Conversation memory enables the chatbot to:

- Maintain conversational context.
- Produce coherent multi-turn interactions.
- Avoid repetitive responses.
- Improve user engagement.
- Simulate real conversational AI systems such as ChatGPT and Gemini.

Only the latest five interactions are included in each prompt to balance contextual understanding with computational efficiency.

```python
# Define Supported Emotions
EMOTIONS = [

    "Joy",

    "Sadness",

    "Love",

    "Hope",

    "Fear",

    "Anger",

    "Loneliness",

    "Excitement",

    "Gratitude",

    "Nostalgia",

    "Confusion",

    "Calm"

]
```

```python
#Create Emotion Detection Prompt
EMOTION_PROMPT = """
You are an emotion classifier.

Identify ONLY ONE dominant emotion from the following list.

Joy
Sadness
Love
Hope
Fear
Anger
Loneliness
Excitement
Gratitude
Nostalgia
Confusion
Calm

Reply with ONLY the emotion name.

User Message:

{}
"""
```

```python
#Create Emotion Detection Function
def detect_emotion(message):

    response = client.models.generate_content(

        model=MODEL_NAME,

        contents=EMOTION_PROMPT.format(message)

    )

    emotion = response.text.strip()

    return emotion
```

```python
#Create Emotion-Based Instructions
EMOTION_STYLE = {

    "Joy":
    "Use bright imagery, sunshine, flowers and celebration.",

    "Sadness":
    "Use gentle comforting imagery and hope.",

    "Love":
    "Use romantic imagery and stars.",

    "Hope":
    "Use sunrise and new beginnings.",

    "Fear":
    "Use courage and light overcoming darkness.",

    "Anger":
    "Transform anger into wisdom and peace.",

    "Loneliness":
    "Comfort the reader using moonlight and stars.",

    "Excitement":
    "Celebrate achievement and dreams.",

    "Gratitude":
    "Express thankfulness using nature.",

    "Nostalgia":
    "Use memories and autumn imagery.",

    "Confusion":
    "Guide gently toward clarity.",

    "Calm":
    "Use peaceful rivers and forests."
}
```

```python
#Update Prompt Builder
def build_prompt(user_message, style="Free Verse"):

    emotion = detect_emotion(user_message)

    emotion_instruction = EMOTION_STYLE.get(

        emotion,

        "Write beautifully."

    )

    prompt = f"""

{PERSONALITY}

Current Emotion

{emotion}

Writing Instruction

{emotion_instruction}

Poetry Style

{style}

User Message

{user_message}

Generate only poetry.

"""

    return prompt
```
